```python
class Kucing:
  # Constructor
  def __init__(self, nama, warna, age):
    self.nama = nama # Attribute
    self.warna = warna # Attribute
    self.age = age # Attribute
  # Method
  def suara(self):
    return f"{self.nama} berkata: Meow!"
# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih",4)
kucing2 = Kucing("Garfield", "Oranye", 5) # Added an age for Garfield
# Mengakses Attribute dan Method
print(kucing1.nama) # Output: Luna
print(kucing1.warna)
print(kucing1.age) # Fixed typo: changed kucing.age to kucing1.age
print(kucing2.suara()) # Output: Garfield berkata: Meow!
```

```python
# Parent Class
class Hewan:
  def makan(self):
    print("Hewan ini sedang makan.")
# Child Class (Mewarisi Hewan)
class Burung(Hewan):
  def terbang(self):
    print("Burung sedang terbang.")
# Object
beo = Burung()
beo.makan() # Mewarisi dari Hewan (Output: Hewan ini sedang makan.)
beo.terbang() # Method milik sendiri
```

```python
class AkunBank:
  def __init__(self, saldo):
    self.__saldo = saldo # Private attribute
  def cek_saldo(self):
    return self.__saldo
  def setor(self, jumlah):
    if jumlah > 0:
      self.__saldo += jumlah
rekening = AkunBank(1000)
rekening.setor(500)
print(rekening.cek_saldo()) # Output: 1500
# print(rekening.__saldo)
```

```python
class Hero:
  def __init__(self, name, hp, attack_power):
    self.name = name
    self.__hp = hp # Make HP private
    self.attack_power = attack_power

  # GETTER: Cara resmi melihat HP
  def get_hp(self):
    return self.__hp

  # SETTER: Cara resmi mengubah HP (dengan validasi)
  def set_hp(self, new_hp_value):
    if new_hp_value < 0:
      self.__hp = 0 # HP tidak boleh negatif
    elif new_hp_value > 1000:
      print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
      self.__hp = 1000
    else:
      self.__hp = new_hp_value

  def info(self):
   # Use get_hp() for consistency with private HP
   print(f"Hero: {self.name} | HP: {self.get_hp()} | Power:{self.attack_power}")

  def serang(self, lawan):
    print(f"{self.name} menyerang {lawan.name}")
    # Assume lawan.diserang uses its own setter/getter for HP
    lawan.diserang(self.attack_power)

  def diserang(self, damage):
    # Update HP using the setter
    sisa_hp = self.get_hp() - damage
    self.set_hp(sisa_hp)
    print(f"{self.name} terkena damage {damage}. Sisa HP: {self.get_hp()}")

class Mage(Hero):
  def __init__(self, name, hp, attack_power, mana):
    super().__init__(name, hp, attack_power)
    self.mana = mana

  def info(self):
    print(f"Name: {self.name} | HP: {self.get_hp()} | Attack Power: {self.attack_power} | Mana: {self.mana}")
    # Mage punya skill khusus
  def skill_fireball(self, lawan):
    if self.mana >= 20:
      print(f"{self.name} menggunakan Fireball ke{lawan.name}!")
      self.mana -= 20
      lawan.diserang(self.attack_power * 2) # Damage 2x lipat
    else:
      print(f"{self.name} gagal skill! Mana tidak cukup.")







```

```python
# ==========================================
# LATIHAN 5: ABSTRACTION & INTERFACE
# ==========================================

from abc import ABC, abstractmethod

# 1. Interface / Abstract Class (KONTRAK)
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
m.info()
h.serang("Serigala")
m.serang("Alucard")

# Menguji coba pembuatan objek dari Abstract Class langsung
try:
    unit = GameUnit()
except TypeError as e:
    print(f"\n[ERROR TERDETEKSI]: {e}")
```

**ANALISA 5**

1. Melanggar Kontrak (Menghapus Method serang)

Pesan Error:

TypeError: Can't instantiate abstract class Hero with abstract method serang

Penjelasan Pesan Error:

Pesan ini menunjukkan bahwa Python menolak pembuatan objek dari class Hero karena class Hero belum melengkapi “kontrak” atau janji yang ditetapkan oleh class induknya, GameUnit. GameUnit sudah menentukan bahwa method serang harus ada. Jadi class Hero masih dianggap abstrak sampai method serang diimplementasikan dengan nyata.

Konsekuensi:

Jika kita lupa membuat method yang dijanjikan di interface, program akan langsung berhenti (error saat runtime). Objek dari class Hero tidak akan pernah bisa dibuat.

2. Mencetak Cetakan (unit = GameUnit())

Alasan Dilarang:

GameUnit adalah Abstract Base Class yang hanya berisi kerangka kosong tanpa kode konkrit (hanya pass). Membuat objek langsung dari GameUnit tidak masuk akal karena unit tersebut tidak memiliki identitas, bentuk, atau perilaku yang jelas.

Kegunaan GameUnit:

Walaupun tidak dapat dijadikan objek nyata, GameUnit tetap sangat berguna sebagai cetakan atau standar (blueprint) dan penjamin konsistensi. Class ini memaksa semua pengembang agar setiap jenis karakter baru—seperti Hero, Monster, NPC, dan sebagainya—memiliki struktur interaksi dan method yang seragam, sehingga tidak ada fitur dasar yang terlewat.

```python
# ==========================================
# LATIHAN 6: POLYMORPHISM
# ==========================================

class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# Tugas Analisis 6.1: Menambahkan class Healer
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")

# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Rafaela") # Ditambahkan ke dalam list
]

print("\n--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

Hasil Tugas Analisis 6:

1. Uji Skalabilitas (Menambahkan Class Healer)

Hasil: Program dapat berjalan dengan sangat lancar tanpa mengalami error sama sekali.

Keuntungan Polimorfisme: Polimorfisme membuat sistem bersifat Scalable (mudah dikembangkan) dan Maintainable (mudah dirawat). Programmer dapat menambahkan puluhan karakter baru di masa depan tanpa perlu mengubah logika utama aplikasi (seperti kode looping pemanggilan serangan). Hal ini menghemat waktu pengembangan dan meminimalkan risiko timbulnya bug pada kode yang sudah ada.

2. Konsistensi Penamaan (tembak_panah pada Archer)

Hasil: Saat method diubah menjadi tembak_panah(), objek Archer tidak lagi menjalankan aksi memanah, melainkan mengeksekusi method serang() bawaan milik class induk (Hero) dengan keluaran teks: "Hero menyerang dengan tangan kosong."

Alasan Pentingnya Kesamaan Nama: Polimorfisme bekerja berdasarkan antarmuka (interface) dan nama method yang seragam. Agar perintah tunggal seperti pahlawan.serang() dapat bekerja secara dinamis pada semua objek, setiap class anak harus melakukan pembaruan (override) menggunakan nama method yang persis sama dengan yang didefinisikan pada class induknya. Jika namanya berbeda, Python akan menganggap method tersebut sebagai fungsi baru yang terpisah, bukan bentuk variasi dari perintah serang().
